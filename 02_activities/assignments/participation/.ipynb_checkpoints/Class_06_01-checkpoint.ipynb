```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0, 100, 50)
```

```python
# show scattered data
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x, y)
```

```python
# create a bar plot
fig, ax = plt.subplots(figsize=(5,3))
ax.bar(x, y)
```

```python
# plot a line
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x, y)
```

```python
#### ACTIVITY
fig, ax = plt.subplots(figsize=(5,3))
ax.hist(y)
```

```python
#### adding labels
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x, y)
ax.set_title('total growth over time')
ax.set_xlabel('time (years)')
ax.set_ylabel('total growth')
fig.tight_layout()
```

```python
font1 = {'family':'sans-serif', 'color':'blue', 'size':20}
font2 = {'family':'monospace','color':'green','size':14}
```

```python
#### changing font
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x, y)
ax.set_title('total growth over time', fontdict = font1)
ax.set_xlabel('time (years)', fontdict = font2)
ax.set_ylabel('total growth', fontdict = font2)
fig.tight_layout()
```

```python
# changing look of data points
fig, ax = plt.subplots(figsize=(5,3))
ax.scatter(x, y, marker = '*', color = 'indigo')
fig.show()
```

```python
# changing look of data points
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x, y, marker = '*', color = 'indigo', linestyle = '--', linewidth = 2)
fig.show()
```

```python
# hex color 
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x, y, marker = '*', color = '#7425b9', linestyle = '--', linewidth = 2)
fig.show()
```

```python
# hex color changes of the markers
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x, y, marker = '*', color = '#7425b9', linestyle = '--', linewidth = 2, markeredgecolor = '#fa9359',markerfacecolor = '#000000')

fig.show()
```

```python
#adding grid lines hex color changes of the markers
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x, y, marker = '*', color = '#7425b9', linestyle = '--', linewidth = 2, markeredgecolor = '#fa9359',markerfacecolor = '#000000')
ax.grid(axis = 'y', color = "blue", linewidth = 2, linestyle = '-.')
fig.show()
```
