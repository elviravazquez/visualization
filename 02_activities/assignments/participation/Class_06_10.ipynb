```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75,50)
x2 = np.array(["Luffy","Zoro","Nami","Usopp" ,"Sanji"])
y2 = np.array([110, 180, 240, 99, 220])
```

```python
#plotting the data in 2 subplots
fig,(ax1, ax2) = plt.subplots(ncols=2,nrows=1,figsize=(7, 3))
ax1.scatter(x1,y1)
ax2.bar(x2,y2)
fig.show()
```

```python
#Activity, modifying the plot
fig,(ax1, ax2) = plt.subplots(ncols=2,nrows=1,figsize=(10, 3))
ax1.scatter(x1,y1, color = 'red')
ax2.bar(x2,y2)

ax1.set_xlabel('x distance', fontsize = 12)
ax1.set_ylabel('y distance', fontsize = 12)
ax2.set_xlabel('Person', fontsize = 12)
ax2.set_ylabel('Number of purchased apples', fontsize = 12)
fig.show()
```

```python
# using a mosaic subplot arrangement
fig, someaxes = plt.subplot_mosaic([['ax1','ax3'],
                                    ['ax2','ax3']],figsize=(7, 4))
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
plt.show()
```

```python
# using a mosaic subplot arrangement
fig, someaxes = plt.subplot_mosaic([['ax1','ax3'],
                                    ['ax2','ax3']],figsize=(7, 4))
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
someaxes["ax1"].set_xlabel('A Big Label',fontsize=16)
someaxes["ax2"].set_xlabel('Another Label',fontsize=16)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious',fontsize=16)
plt.show()
```

```python
# changinig font size of the graph using the layout argument
fig, someaxes = plt.subplot_mosaic([['ax1','ax3'],
                                    ['ax2','ax3']],figsize=(7, 4),
                                    layout = 'constrained')
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
someaxes["ax1"].set_xlabel('A Big Label',fontsize=16)
someaxes["ax2"].set_xlabel('Another Label',fontsize=16)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious',fontsize=16)
plt.show()
```

```python
x = np.array(["Luffy","Zoro","Nami","Usopp" ,"Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])

fig, ax = plt.subplots(figsize=(7, 3))

ax.bar(x, y1, color = "indigo")
ax.plot(x, y2, color = "red")
ax.set_xlabel('names')
```

```python
# adding error information

y2_sd = np.std(y2) # calculating std function

fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")

#adding error bars
ax.errorbar(x, #our x values
            y2, #our y values
            yerr = y2_sd,
            fmt = "none")
```

```python
#customizing errorbar appearance
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")
#adding error bars
ax.errorbar(x, #our x values
            y2, #our y values
            yerr = y2_sd,
            fmt = "none", 
           ecolor = 'indigo',
           elinewidth = 4,
           capsize = 6, 
           capthick = 4)
```

```python
#skipping every other error bar
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")
#adding error bars
ax.errorbar(x, #our x values
            y2, #our y values
            yerr = y2_sd,
            fmt = "none", 
           ecolor = 'indigo',
           elinewidth = 4,
           capsize = 6, 
           capthick = 4, 
           errorevery = 2)
```

```python
# add images to plots

from PIL import Image # to open images
import requests # to get images from URLs
from io import BytesIO # to store images
```

```python
url = "https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png"
headers = {"User-Agent": "Mozilla/5.0","Accept": "image/*,*/*;q=0.8","Referer": "https://en.wikipedia.org/"}

response = requests.get(url, headers=headers)
image_file = BytesIO(response.content)
image = Image.open(image_file)
```

```python
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")

ax_image = fig.add_axes([0.1, # x coordinate (ON FIGURE, NOT AXES)
                        0.11, # y coordinate (ON FIGURE, NOT AXES)
                        0.15, # image width
                        0.35]) # image height
ax_image.imshow(image)
ax_image.axis('off') # to remove the axis/box of the image
```
