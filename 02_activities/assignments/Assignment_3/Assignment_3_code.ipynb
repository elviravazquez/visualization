```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
# upload the data
data = pd.read_csv('dataset.csv')
```

```python
# printing the data, FSA corresponds to the first digits of the postal code
data 
```

```python
''' First plot, number of dog and cats in a given year'''


fig, ax = plt.subplots(figsize=(7, 3))
# setting the palette and style of the plot
sns.set_theme(style = 'white', palette="colorblind")

#plotting the bar plot
sns.countplot(data, x="Year", hue="ANIMAL_TYPE", palette=["m", "g"])

#adding axes labels
ax.set_xlabel('Year', fontsize = 14)
ax.set_ylabel('Count', fontsize = 14)
ax.set_title('Dogs/Cats vs Year', fontsize = 14)

fig.text(0.03, -0.12,"This graph was last updated on Jun 14, 2026",ha='left', color = 'black', size = 13)
fig.text(0.03, -0.18,"Source: City of Toronto",ha='left', color = 'gray', size = 13)

plt.savefig('Figure1.png', bbox_inches='tight', dpi= 300)

plt.show()
```

```python
all_dogs = data[data['ANIMAL_TYPE'] == 'DOG'] # selecting all the dogs from the dataframe
all_dogs_2025 = all_dogs[all_dogs['Year']==2025] # selecting all the dogs from 2025
```

```python
''' Second plot, Top 5 most popular dog breeds in 2025, looking at the data, real one is below'''

fig, ax = plt.subplots(figsize=(10, 3))
# setting the palette and style of the plot
sns.set_theme(style = 'white', palette="colorblind")

#finding the top 10 breeds for simplicity
top_breeds = all_dogs_2025['PRIMARY_BREED'].value_counts().head(5).index


#plotting the bar plot
sns.countplot(all_dogs_2025[all_dogs_2025['PRIMARY_BREED'].isin(top_breeds)], x="PRIMARY_BREED")

#adding axes labels
ax.set_xlabel('Year', fontsize = 14)
ax.set_ylabel('Count', fontsize = 14)
ax.set_title('Top 5 most popular dog breeds in 2025', fontsize = 15)

fig.text(0.25, -0.3,"This graph was last updated on Jun 14, 2026",ha='center', color = 'black', size = 13)
plt.xticks(rotation=30, ha='center') # rotating the tick label
plt.ylim(0, 6000)

plt.savefig('Figure1.png', bbox_inches='tight', dpi= 300)


plt.show()
```

```python
from PIL import Image # to open images
import requests # to get images from URLs
from io import BytesIO # to store images
```

```python
### getting all the pictures of the dogs for the visualization

#adding figure of labrador retriever :)
url1 = "https://pngimg.com/uploads/labrador_retriever/labrador_retriever_PNG49.png"

response1 = requests.get(url1)
image_file1 = BytesIO(response1.content)
image1 = Image.open(image_file1)

#adding figure of golden retriever :)
url2 = "https://www.pngarts.com/files/3/Golden-Retriever-PNG-Pic.png"

response2 = requests.get(url2)
image_file2 = BytesIO(response2.content)
image2 = Image.open(image_file2)

# adding image of german shepherd
url3 = "https://www.nicepng.com/png/full/379-3793572_dog-insurance-german-shepherd.png"

response3 = requests.get(url3)
image_file3 = BytesIO(response3.content)
image3 = Image.open(image_file3)

# adding image of shih tzu
url4 = "https://png.pngtree.com/png-vector/20250222/ourmid/pngtree-clipart-of-a-shih-tzu-dog-with-no-background-png-image_15552619.png"

response4 = requests.get(url4)
image_file4 = BytesIO(response4.content)
image4 = Image.open(image_file4)

# adding image of cocker spaniel
url5 = "https://www.dogstrust.org.uk/images/600x450/assets/2023-05/cockerspaniel_stock.png"

response5 = requests.get(url5)
image_file5 = BytesIO(response5.content)
image5 = Image.open(image_file5)
```

```python
''' Second plot, Top 5 most popular dog breeds in 2025'''


fig, ax = plt.subplots(figsize=(10, 4))
# setting the palette and style of the plot
sns.set_theme(style = 'white', palette="colorblind")

#finding the top 10 breeds for simplicity
top_breeds = all_dogs_2025['PRIMARY_BREED'].value_counts().head(5).index


#plotting the bar plot
sns.countplot(all_dogs_2025[all_dogs_2025['PRIMARY_BREED'].isin(top_breeds)], x="PRIMARY_BREED", 
             order = top_breeds)

#adding axes labels
ax.set_xlabel('Year', fontsize = 14)
ax.set_ylabel('Count', fontsize = 14)
ax.set_title('Top 5 most popular dog breeds in 2025 in Toronto', fontsize = 15)

fig.text(0.1, -0.3,"This graph was last updated on Jun 14, 2026",ha='left', color = 'black', size = 13)
fig.text(0.1, -0.35,"Source: City of Toronto",ha='left', color = 'gray', size = 13)
plt.xticks(rotation=30, ha='center') # rotating the tick label
plt.ylim(0, 4700)

# adding labrador retriever image
ax_image1 = fig.add_axes([0.14, # x coordinate (ON FIGURE, NOT AXES)
                        0.1, # y coordinate (ON FIGURE, NOT AXES)
                        0.13, # image width
                        0.35]) # image height
ax_image1.imshow(image1)
ax_image1.axis('off') # to remove the axis/box of the image

#adding golden retriever image
ax_image2 = fig.add_axes([0.29, # x coordinate (ON FIGURE, NOT AXES)
                        0.11, # y coordinate (ON FIGURE, NOT AXES)
                        0.15, # image width
                        0.35]) # image height
ax_image2.imshow(image2)
ax_image2.axis('off') # to remove the axis/box of the image

# adding german shepherd image
ax_image3 = fig.add_axes([0.44, # x coordinate (ON FIGURE, NOT AXES)
                        0.11, # y coordinate (ON FIGURE, NOT AXES)
                        0.15, # image width
                        0.35]) # image height
ax_image3.imshow(image3)
ax_image3.axis('off') # to remove the axis/box of the image

# adding shih tzu image
ax_image4 = fig.add_axes([0.61, # x coordinate (ON FIGURE, NOT AXES)
                        0.10, # y coordinate (ON FIGURE, NOT AXES)
                        0.12, # image width
                        0.30]) # image height
ax_image4.imshow(image4)
ax_image4.axis('off') # to remove the axis/box of the image

# adding cocker spaniel image
ax_image5 = fig.add_axes([0.73, # x coordinate (ON FIGURE, NOT AXES)
                        0.047, # y coordinate (ON FIGURE, NOT AXES)
                        0.2, # image width
                        0.4]) # image height
ax_image5.imshow(image5)
ax_image5.axis('off') # to remove the axis/box of the image

plt.savefig('Figure2.png', bbox_inches='tight', dpi= 300)


plt.show()
```
